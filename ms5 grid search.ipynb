```python
import numpy as np
from grid_search import evaluate_algorithm, grid_search_algorithm
from brute_force import find_phi_brute_force, find_phi_fixed_budget_brute_force
from linear_search import find_phi_linear_search, find_phi_fixed_budget_linear_search
from binary_search import find_phi_binary_search, find_phi_binary_search_anneal_m, find_phi_fixed_budget_binary_search, find_phi_fixed_budget_binary_search_anneal_m
from reverse_engineering import find_phi_reverse_engineering, find_phi_fixed_budget_reverse_engineering

rng = np.random.default_rng(42)
R = 1000
```
