# QM when no upper bound is known

```python
import numpy as np
from grid_search import evaluate_algorithm_single, grid_search_algorithm_single
from brute_force import find_phi_brute_force, find_phi_fixed_budget_brute_force
from linear_search import find_phi_linear_search, find_phi_fixed_budget_linear_search
from binary_search import find_phi_binary_search, find_phi_binary_search_anneal_m, find_phi_fixed_budget_binary_search, find_phi_fixed_budget_binary_search_anneal_m, find_phi_fixed_budget_binary_search_delayed
from reverse_engineering import find_phi_reverse_engineering, find_phi_fixed_budget_reverse_engineering

rng = np.random.default_rng(42)
R = 1000
```

```python
phi_max = np.pi/2
print(f"{phi_max = }")
phi_min = 0.01

params = {
    "budget": 10_000,
}

results = evaluate_algorithm_single(
    find_phi_fn=find_phi_fixed_budget_brute_force,
    params=params,
    R=R,
    phi_max=phi_max,
    phi_min=phi_min,
    eps=1e-3,
    rng=rng,
)
```

```python
phi_max = np.pi/2
print(f"{phi_max = }")
phi_min = 0.01

N_min = max(np.pi//(2*phi_max), 1)
N_max = max(np.pi//(2*phi_min), 1)

print(N_min, N_max)

print("""
      -----------------------------------------------
      -------------Linear search---------------------
      -----------------------------------------------""")

param_grid = {
    "m_exploration": np.logspace(1, 4, num=50, dtype=int),
    "lookback_window": [1, 2, 5],
    "safeguard": [0, 1, 2, 5],
    "budget": [10_000],
    "inc": [1, 2, 5, 10]
}

results = grid_search_algorithm_single(
    find_phi_fn=find_phi_fixed_budget_linear_search,
    param_grid=param_grid,
    R=R,
    phi_min=phi_min,
    phi_max=phi_max,
    eps=1e-3,
    rng=rng,
    lambda_budget=0,
)

print("""
      -----------------------------------------------
      -------------Binary serach---------------------
      -----------------------------------------------""")

param_grid = {
    "m_exploration": np.logspace(1, 4, num=50, dtype=int),
    "safeguard": [0, 1, 2, 5],
    "budget": [10_000],
    "conf": [0.5, 0.8, 0.9]
}

results = grid_search_algorithm_single(
    find_phi_fn=find_phi_fixed_budget_binary_search,
    param_grid=param_grid,
    R=R,
    phi_min=phi_min,
    phi_max=phi_max,
    eps=1e-3,
    rng=rng,
    lambda_budget=0,
)

print("""
      -----------------------------------------------
      -------------Binary serach (anneal m)----------
      -----------------------------------------------""")

param_grid = {
    "m_exploration": np.logspace(1, 4, num=50, dtype=int),
    "safeguard": [0, 1, 2, 5],
    "budget": [10_000],
    "conf": [0.5, 0.8, 0.9],
    "max_b_steps_sub": [2, 3, 4],
    "delta": [3, 4, 5],
}

results = grid_search_algorithm_single(
    find_phi_fn=find_phi_fixed_budget_binary_search_anneal_m,
    param_grid=param_grid,
    R=R,
    phi_min=phi_min,
    phi_max=phi_max,
    eps=1e-3,
    rng=rng,
    lambda_budget=0,
)

print("""
      -----------------------------------------------
      -------------Binary serach (delayed)----------
      -----------------------------------------------""")

param_grid = {
    "m_exploration": np.logspace(1, 4, num=50, dtype=int),
    "m_reference": np.logspace(1, 4, num=50, dtype=int),
    "safeguard": [0, 1, 2, 5],
    "budget": [10_000],
    "conf": [0.5, 0.8, 0.9],
}

results = grid_search_algorithm_single(
    find_phi_fn=find_phi_fixed_budget_binary_search_delayed,
    param_grid=param_grid,
    R=R,
    phi_min=phi_min,
    phi_max=phi_max,
    eps=1e-3,
    rng=rng,
    lambda_budget=0,
)

print("""
      -----------------------------------------------
      -------------Reverse Engineering --------------
      -----------------------------------------------""")

param_grid = {
    "m_exploration": np.logspace(1, 4, num=50, dtype=int),
    "safeguard": [0.8, 0.9, 0.95],
    "budget": [10_000],
}

results = grid_search_algorithm_single(
    find_phi_fn=find_phi_fixed_budget_reverse_engineering,
    param_grid=param_grid,
    R=R,
    phi_min=phi_min,
    phi_max=phi_max,
    eps=1e-3,
    rng=rng,
    lambda_budget=0,
)
```
