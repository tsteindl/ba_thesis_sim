```python
import numpy as np
from sim import simulate_errors
from grid_search import evaluate_algorithm_single


rng = np.random.default_rng(42)
R=1000

phi_max = 0.1
print(f"{phi_max = }")
phi_min = 0.01


def find_phi_fixed_budget_separable(rng, phi, phi_max, phi_min, budget):
    # N_min = max(np.pi//(2*phi_max), 1)
    N = 1
    m = int(budget/N)
    phi_hat = simulate_errors(rng, phi, m, N)
    budget_used = m * N
    
    return phi_hat, budget_used


params = {
    "budget": 10_000,
}

results = evaluate_algorithm_single(
    find_phi_fn=find_phi_fixed_budget_separable,
    params=params,
    R=R,
    phi_max=phi_max,
    phi_min=phi_min,
    eps=1e-3,
    rng=rng,
)
```
