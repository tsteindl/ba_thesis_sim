# Binary search with history

```python
from sim import simulate_errors

def find_phi_binary_search_hist(rng, phi, phi_max, phi_min, m_exploration=400, m_exploitation=1_000, lookback_window=5, safeguard=1):
    N_min = max(np.pi//(2*phi_max), 1)
    N_max = np.pi//(2*phi_min)
    N = N_min
    
    lb = N_min
    ub = N_max
    # N += (ub - N) // 2
    
    budget_used = 0
    r_mean = 0
    phi_hat_list = []
    counter = 0
    states = []
    
    done = False
    while not done:
        phi_hat = simulate_errors(rng, phi, m_exploration, N)
        budget_used += m_exploration * N
        phi_hat_list.append(phi_hat)
        
        r_mean_new = np.mean(phi_hat_list)
        if len(phi_hat_list) >= 2 and r_mean_new < r_mean:
            counter += 1
        else:
            counter = 0
            
        r_mean = r_mean_new
        
        states.append((N, lb, ub))
        
        if counter >= lookback_window:
            N = states[-lookback_window][0]
            lb = states[-lookback_window - 1][1]
            ub = N
            N -= (N - lb) // 2
            
            counter = 0
        elif N >= N_max:
            done = True
        else:
            N += (ub - N) // 2
            lb = N
        
        
        if states[-1][0] == N or N < N_min or N > N_max:
            done = True
    
    N = max(1, N - safeguard) # to avoid overshooting
    
    phi_hat = simulate_errors(rng, phi, m_exploitation, N)
    budget_used += m_exploitation * N
    
    return phi_hat, budget_used
```

```python
find_phi_binary_search_hist(rng, 0.05, 0.1, 0.01, 400, 1000, 2, 1)
```

```python
params = {
  "m_exploration": 120,
  "m_exploitation": 3393,
#   "conf": 0.5,
  "safeguard": 2,
}

results = evaluate_algorithm(
    find_phi_fn=find_phi_binary_search_hist,
    params=params,
    R=R,
    phi_max=0.1
    phi_min=0.01,
    eps=1e-3,
    rng=rng,
)
```
