# QM when no upper bound is known

$$
\phi\sim\mathcal U(0.01, \pi/2)
$$

```python
import numpy as np
from grid_search import evaluate_algorithm
from brute_force import find_phi_brute_force
from linear_search import find_phi_linear_search
from binary_search import find_phi_binary_search, find_phi_binary_search_anneal_m
from reverse_engineering import find_phi_reverse_engineering

rng = np.random.default_rng(42)
R = 1000
```

```python
params = {
    "m": 4_500,
}

results = evaluate_algorithm(
    find_phi_fn=find_phi_brute_force,
    params=params,
    R=R,
    # phi_max=np.pi/2,
    phi_max=2*np.pi,
    phi_min=0.01,
    eps=1e-3,
    rng=rng,
)
```
